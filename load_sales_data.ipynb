What are database relationships?
A database relationship is a connection between two or more tables in a database.

- One-to-one (1:1) one record in one table is linkes to a single record in another table.
- One-to-many (1:M) a single record in one table can be related to multiple records in another table.
- Many-to-many (M:M) records in one table can be related to multiple records in another table.

```python
# sqldf accepts 2 parameters: a SQL query string, and a set of environment variables (locals() or globals()
#) which is optional and can be omitted

pysqldf = lambda q: sqldf(q, globals())

#Any pandas data frames will be automatically detected by pandasql
#We can query any pandas DataFrame using SQL in the same way as we extract data from any table using SQL.
```

```python
# the SELECT statement is used to retrieve data from a database
# we cna choose specific columns or all columns

# the WHERE is used to filter records based on conditions

pysqldf("SELECT Customer_ID, Customer_Name FROM customers WHERE Customer_ID IS NOT NULL")
```

```python
pysqldf("SELECT Customer_ID, Customer_Name FROM customers WHERE Age >= 70")
```

```python
# AND, OR, NOT operators allow to combine multiple conditions in the WHERE clause
pysqldf("SELECT Customer_ID, Customer_Name, Age FROM customers WHERE Age > 70 AND Gender = 'male'")
```

```python
# The BETWEEN operator selects values within a range. It is inclusive of the boundary values

pysqldf("SELECT Customer_ID, Customer_Name, Age FROM customers WHERE Age BETWEEN 75 AND 99 ")
```

```python
# The IN operator allows to specify multiple value sin a WHERE clause
pysqldf("""SELECT Customer_ID, Customer_Name, Age, Customer_Tier FROM customers WHERE Customer_Tier IN ('Gold', 'Platinum') 
AND Age > 73""")
```

To use directly sql commands we need to create a database, not a dataframe pandas.

```python
import sqlite3

# To interact with an SQLite database, we first need to establish a connection to 
# it using the connect() method

import sqlite3

# conn is the connection to the database object (database.db)
conn = sqlite3.connect("database.db")

customers.to_sql("customers", conn, index=False, if_exists="replace")
products.to_sql("products", conn, index=False, if_exists="replace")
sales.to_sql("sales", conn, index=False, if_exists="replace")

conn.close()

%sql sqlite:///database.db
```

```python
%%sql
SELECT Customer_ID, Customer_Name, Age, Customer_Tier FROM customers 
WHERE Customer_Tier IN ('Gold', 'Platinum')
AND Age > 74
```

```python
%%sql
-- The LIKE operator searches for a specified pattern
-- % = any number of characters
-- _ = a single character
-- Select customers whose name contains "ali"

SELECT Customer_ID, Customer_Name, Age
FROM customers
WHERE Customer_Name LIKE '%alid%'
AND Age in (45,60)
```

```python
%%sql
-- The ORDER BY clause is used to sort the result set in ascending or descending order
-- by default, it sorts in ascending order
-- ASC: ascending, DESC descending

SELECT Customer_Name, Age FROM customers WHERE Age in (19,22) AND
Customer_Name LIKE '%priya%'
ORDER BY Age DESC
```

```python
%%sql
--perform calculations on a set of values
-- SUM returns the total sum of a numeric columns, AVG the average
-- COUNT returns the number of rows of non-null values

SELECT SUM(Selling_Price), AVG(Selling_Price), COUNT(Selling_Price) 
FROM products WHERE Category = 'Electronics'
```

```python
%%sql
SELECT SUM(Selling_Price) AS total_price FROM products;
```

```python
%%sql
SELECT AVG(Selling_Price) AS avg_price FROM products
```

```python
%%sql

-- the GROUP BY statement is used to group rows that have the same values in specific columns

SELECT Category, SUM(Selling_Price) AS 'tot selling price'
FROM products
GROUP BY Category
```

```python
result = pysqldf(""" SELECT Category, SUM(Selling_Price) AS 'tot selling price'
FROM products
GROUP BY Category """)

result.head(3)
```

```python
%%sql

-- the HAVING clause is used to filter records after grouping,
-- similar to WHERE; but HAVING operates on aggregated data

SELECT Category, SUM(Selling_Price) AS 'tot selling price'
FROM products
GROUP BY Category
HAVING Category in ('Beauty', 'Books')
```

```python
%%sql

SELECT Category, SUM(Selling_Price) AS tot_selling_price
FROM products
GROUP BY Category
HAVING tot_selling_price < 220000
```

```python
%%sql

-- the DISTINCT keyword is used to return only unique values

SELECT DISTINCT Category from products
```

## SQL ARITHMETIC

```python
%%sql

-- ex to subtract 10% of sale

SELECT Product_Name, Selling_Price, (Selling_Price*0.1) AS Sale, Selling_Price-(Selling_Price*0.1) AS selling_price_sale 
FROM products
WHERE Sale > 12000
```

```python
%%sql
-- There are also CEIL, FLOOR, ABS, ROUND commands

SELECT Product_Name, Selling_Price, ROUND(Selling_Price*0.1, 2) AS Sale, ROUND(Selling_Price-(Selling_Price*0.1),2) 
AS selling_price_sale 
FROM products
WHERE Sale > 12000
```

## NULL values

```python
%%sql

-- there are also functions like IS NULL; IS NOT NULL, COALESCE() to handle null values

SELECT * FROM products WHERE Selling_Price IS NULL
```

```python
%%sql

-- returns 0 if the price is NULL

SELECT COALESCE(Selling_Price, 0) FROM products
WHERE Selling_Price > 500000
```
